```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
sns.set_theme(style = "dark")

sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    size = "size",
    style = "smoker"
)

```

```python
xval = [1,2,3,4,5]

yval = [i**2 for i in xval]

sns.set_theme(style = "ticks")
sns.relplot(
    x = xval,
    y = yval,
    kind = "line")
```

```python
# SCATER PLOT

tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time")
```

```python
# LINE PLOT

flights = sns.load_dataset("flights")

sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers")
```

```python
# BAR PLOTS

tips = sns.load_dataset("tips")

sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex")
```

```python
# BOX PLOT

tips = sns.load_dataset("tips")

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
    )
```

```python
# HISTOGRAM 

penguins = sns.load_dataset("penguins")

sns.histplot(
    data =penguins,
    x = "body_mass_g",
    bins = 19
    )
```

```python
# HEAT MAP

flights_pivot = flights.pivot(index = "month", columns = "year", values = "passengers")

sns.heatmap(
    flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "g")
    
    
```

```python
# BEST PRACTICE 

fig, ax = plt.subplots()

sns.lineplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    marker = "o",
    errorbar=None,
    ax = ax)
    
```
